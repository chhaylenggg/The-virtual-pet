Virtual pet
- The Goal: Create a Pet class with attributes like name, hunger, energy, and happiness. Create methods like feed(), play(), and sleep() that alter these attributes.
- Concepts Tested: Classes, __init__, self, instance attributes, methods.

```python
class Pet:
    def __init__(self, name):
        self.name = name
        self.hunger = 5
        self.energy = 5
        self.happiness = 5
    def feed(self):
        print (f"{self.name} has been fed")
        self.hunger -= 1
    def play(self):
        print (f"{self.name} is playing")
        self.happiness += 1
        self.energy -= 1
    def sleep(self):
        print (f"{self.name} is sleeping")
        self.energy += 1
    def status(self):
        print(f"""
Name : {self.name}
Hunger : {self.hunger}
Energy : {self.energy}
Happiness : {self.happiness}
        """)

pet = Pet("luffy")
pet.feed()
pet.play()
pet.sleep()
pet.status()
```
